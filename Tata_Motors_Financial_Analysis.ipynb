# **Revenue Growth**





```python
pnl["revenue_growth_%"]=pnl["Revenue"].pct_change()*100

pnl[["Financial Year","Revenue","revenue_growth_%"]]
```

# PAT Growth

```python
pnl["PAT_Growth_%"]=pnl["PAT"].pct_change()*100

pnl[['Financial Year','Revenue','revenue_growth_%',"PAT","PAT_Growth_%"]]
```

# **Operating Profit Margin**

```python
pnl["Operating_margin_%"]=(pnl["Operating Profit"]/pnl["Revenue"]*100)

pnl[["Financial Year","Revenue","Operating Profit","Operating_margin_%"]]
```

# **NET PROFIT**

```python
pnl["netprofit_margin_%"]=(pnl["PAT"]/pnl["Revenue"]*100)

pnl[["Financial Year","Revenue","PAT","netprofit_margin_%"]]
```

# **Current Ratio**

```python
bs["current_ratio"]=(bs["Current Assets"]/bs["Current Liabilities"]*100)

bs[["Financial Year","Current Assets","Current Liabilities","current_ratio"]]
```

# **Return on Equity (ROE)**

```python
bs["ROE_%"] = (pnl["PAT"] / bs["Total Equity"]) * 100

bs[["Financial Year","Total Equity","ROE_%"]]
```

# **Debt-to-Equity Ratio**

```python
bs["Debt_to_Equity"] = (bs["Total Debt"] / bs["Total Equity"])

bs[["Financial Year","Total Debt","Total Equity","Debt_to_Equity"]]
```

# **Return on Assets (ROA)**

```python
bs["ROA_%"] = (pnl["PAT"] / bs["Total Assets"]) * 100

bs[["Financial Year","Total Assets","ROA_%"]]
```

## **Operating Cash Flow to PAT**

```python
cash_flow["Ocf_to_PAT"]=(cash_flow["Operating Cash Flow"]/pnl["PAT"])

cash_flow[["Financial Year","Operating Cash Flow","Ocf_to_PAT"]]
```

# **Cash Flow Margin**

```python
cash_flow["cash_flow_margin_%"]=(cash_flow["Operating Cash Flow"]/pnl["Revenue"]*100)

cash_flow[["Financial Year","Operating Cash Flow","cash_flow_margin_%"]]
```

# **Revenue Trend**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(pnl["Financial Year"],pnl["Revenue"],marker="*")

plt.title("Tata Motors Revenue Trend (FY2021–FY2025)")
plt.xlabel("Financial Year")
plt.ylabel("Revenue (₹ crore)")

plt.grid(True)
plt.show()
```

# **Combined KPI table**

```python
analysis = pd.DataFrame({
    "Financial Year": pnl["Financial Year"],
    "Revenue": pnl["Revenue"],
    "Revenue Growth %": pnl["revenue_growth_%"],
    "Operating Margin %": pnl["Operating_margin_%"],
    "Net Profit Margin %": pnl["netprofit_margin_%"],
    "Debt to Equity": bs["Debt_to_Equity"],
    "Current Ratio": bs["current_ratio"],
    "ROE %": bs["ROE_%"],
    "ROA %": bs["ROA_%"],
    "Operating Cash Flow": cash_flow["Operating Cash Flow"],
    "Cash Flow Margin %": cash_flow["cash_flow_margin_%"]
})

analysis
```

# **Find the strongest and weakest years**

```python
print("Highest Revenue:")
print(analysis.loc[analysis["Revenue"].idxmax(), ["Financial Year", "Revenue"]])

print("\nHighest Operating Margin:")
print(analysis.loc[analysis["Operating Margin %"].idxmax(),
                   ["Financial Year", "Operating Margin %"]])

print("\nLowest Debt to Equity:")
print(analysis.loc[analysis["Debt to Equity"].idxmin(),
                   ["Financial Year", "Debt to Equity"]])

print("\nHighest Operating Cash Flow:")
print(analysis.loc[analysis["Operating Cash Flow"].idxmax(),
                   ["Financial Year", "Operating Cash Flow"]])
```

# **Calculate the overall 5-year change**

```python
start = analysis.iloc[0]
end = analysis.iloc[-1]

print("Revenue change:", end["Revenue"] - start["Revenue"])
print("Operating Margin change:",
      end["Operating Margin %"] - start["Operating Margin %"])
print("Debt-to-Equity change:",
      end["Debt to Equity"] - start["Debt to Equity"])
print("Current Ratio change:",
      end["Current Ratio"] - start["Current Ratio"])
print("ROE change:",
      end["ROE %"] - start["ROE %"])
print("ROA change:",
      end["ROA %"] - start["ROA %"])
print("Operating Cash Flow change:",
      end["Operating Cash Flow"] - start["Operating Cash Flow"])
```
